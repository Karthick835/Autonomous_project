```
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

# Load Dataset
df = pd.read_csv('C:/Users/admin/Desktop/llm and ml project/uploads/c248b784_india_crime_combined_2001_2024.csv')
print("Dataset Loaded Successfully. Shape:", df.shape)
df.head()
```

```
# Plot Correlation Matrix Heatmap for Numerical Features
numeric_cols = df.select_dtypes(include=[np.number]).columns
if len(numeric_cols) >= 2:
    plt.figure(figsize=(10, 6))
    sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
    plt.title("Numerical Feature Correlation Matrix", fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()
```

```
# Plot Target Variable Distribution
plt.figure(figsize=(8, 4))
if pd.api.types.is_numeric_dtype(df['year']) and df['year'].nunique() > 10:
    sns.histplot(df['year'].dropna(), kde=True, color='#6366f1')
    plt.title("Target Distribution: year", fontsize=13, fontweight='bold')
else:
    sns.countplot(x='year', data=df, palette='viridis')
    plt.title("Target Class Distribution: year", fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```
